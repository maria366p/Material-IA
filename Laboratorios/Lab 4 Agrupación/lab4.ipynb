¿25 faltantes es mucho o poco?

```python
(25 / len(data)) * 100
```

Solo el 0.25% del dataset tiene valores faltantes. La imputación contextualizada resulta adecuada debido a la relación operacional existente entre el retraso de salida y el retraso de llegada. Desde el dominio del problema, un vuelo que despega con retraso tiende también a llegar tarde, por lo que ambas variables presentan dependencia estadística y coherencia operacional. Utilizar esta relación permite preservar mejor la estructura del dataset y reducir el sesgo introducido por imputaciones simples basadas únicamente en medidas de tendencia central.
